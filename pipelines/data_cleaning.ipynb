```python
import pandas as pd 
import numpy as np 
```

```python
df = pd.read_csv("../data/customer_support_tickets.csv")
df
```

renaming the columns to the snake_case

```python
df.rename(columns=lambda x: x.strip().lower().replace(' ', '_'), inplace=True)
df
```

replacing the {product_purchased} in the ticket description to the actual values

```python
# for idx in df.index:
#     ticket_description = df.loc[idx, "ticket_description"]
#     replace_value = df.loc[idx, "product_purchased"]
#     df.loc[idx, "ticket_description"] = ticket_description.replace("{product_purchased}", replace_value)

# this was my way but turns out there is a better way 

df["ticket_description"] = df.apply(
    lambda row: row["ticket_description"].replace("{product_purchased}", str(row["product_purchased"])), 
    axis= 1
)
```

```python
# Strip or replace non-ASCII / problematic characters from all text columns
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].astype(str).str.encode('ascii', 'ignore').str.decode('ascii')
```

```python
df.to_csv("../data/Cleaned_Data.csv", index=False)
```
